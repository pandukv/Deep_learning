```python
import numpy as np
```

```python
def one_hot_encode_words(text):
    """One-hot encodes words in a given text."""

    words = text.lower().split()  # Tokenize and lowercase
    unique_words = sorted(list(set(words)))
    num_words = len(unique_words)

    word_to_index = {word: i for i, word in enumerate(unique_words)}
    encoded_text = np.zeros((len(words), num_words), dtype=int)

    for i, word in enumerate(words):
        encoded_text[i, word_to_index[word]] = 1

    return encoded_text, unique_words, word_to_index
```

```python
def one_hot_encode_characters(text):
    """One-hot encodes characters in a given text."""

    text = text.lower() # lowercase
    unique_chars = sorted(list(set(text)))
    num_chars = len(unique_chars)

    char_to_index = {char: i for i, char in enumerate(unique_chars)}
    encoded_text = np.zeros((len(text), num_chars), dtype=int)

    for i, char in enumerate(text):
        encoded_text[i, char_to_index[char]] = 1

    return encoded_text, unique_chars, char_to_index
```

```python
# Example usage (words):
text_words = "This is a simple example. This example is simple."
encoded_words, unique_words, word_to_index = one_hot_encode_words(text_words)

print("Word-level one-hot encoding:")
print("Encoded text:\n", encoded_words)
print("\nUnique words:", unique_words)
print("\nWord to index mapping:", word_to_index)

# Example usage (characters):
text_chars = "Hello, world!"
encoded_chars, unique_chars, char_to_index = one_hot_encode_characters(text_chars)

print("\nCharacter-level one-hot encoding:")
print("Encoded text:\n", encoded_chars)
print("\nUnique characters:", unique_chars)
print("\nCharacter to index mapping:", char_to_index)
```
